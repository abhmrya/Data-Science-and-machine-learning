```python

# Create an Index
idx1 = pd.Index([1, 2, 3, 4, 5])
idx2 = pd.Index([3, 4, 5, 6, 7])
```

```python
print("Append:", idx1.append(idx2))
```

```python
idx1
```

```python
print("Diff:", idx1.diff(idx2))
```

```python
print("Intersection:", idx1.intersection(idx2))
```

```python
print("Union:", idx1.union(idx2))
```

```python
print("Isin:", idx1.isin([2, 5]))
```

```python
idx1
```

```python
print("Delete:", idx1.delete(2))   # remove element at position 2 (zero-based)
```

```python
print("Drop:", idx1.drop([2, 4]))
```

```python
print("Insert:", idx1.insert(2, 99))
```

```python
print("Is Monotonic:", idx1.is_monotonic)
```

```python
print("Is Unique:", idx1.is_unique)
```

```python
print("Unique:", idx1.unique())
```
