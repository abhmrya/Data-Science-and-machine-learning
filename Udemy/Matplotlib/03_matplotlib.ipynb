```python
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
import numpy as np
x = np.linspace(0,5,10)
y = x**2
```

```python
fig = plt.figure()

ax = fig.add_axes([0,0,1,1])
# ax.plot(x,y,color = 'green')
ax.plot(x,y,color = "#0F1477",linewidth = 3,alpha = 1,linestyle = '-',marker = '*',markersize=40,
        markerfacecolor='red',markeredgewidth=3,markeredgecolor='green')
ax.set_ylim([0,1])
ax.set_ylim([0,2])
```
